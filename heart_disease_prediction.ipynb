```python
#Libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
#Loading Data
data = pd.read_csv('/kaggle/input/datasets/guninder07/heart-disease/heart_disease_data.csv')
data.head()
data.shape
```

```python
data.info()
```

```python
#checking for missing values
data.isnull().sum()
```

```python
#Splitting features and target
x = data.drop(columns = 'target')
y = data['target']
x,y
```

```python
#Spliting the dataser
x_train, x_val, y_train, y_val = train_test_split(x, y, test_size=0.25, random_state=42)
```

```python
#Implementing Logistic Regression on training data
log_reg = LogisticRegression(max_iter = 1000)
log_reg.fit(x_train, y_train)
log_reg_pred = log_reg.predict(x_train)


#accuracy
acc = accuracy_score(log_reg_pred, y_train)
acc
```

```python
#now for test data
log_reg_pred_test = log_reg.predict(x_val)


#accuracy
acc = accuracy_score(log_reg_pred_test, y_val)
acc
```
